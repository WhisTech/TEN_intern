# numpy_vector_calculations

```python
import numpy as np

# Create one-dimensional arrays
a = np.array([10, 20, 30, 40])
b = np.array([5, 10, 15, 20])

print("Array A:", a)
print("Array B:", b)

print("Dimensions:", a.ndim)
print("Shape:", a.shape)
print("Data type:", a.dtype)
```

```python
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
vector_a = np.array([2, 3, 4])
vector_b = np.array([5, 6, 7])

result = np.dot(vector_a, vector_b)

print("Dot Product:", result)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

print("Matrix Addition:\n", A + B)

print("Matrix Multiplication:\n", np.dot(A, B))

print("Transpose of A:\n", A.T)
```

```python
# Original vector
vector = np.array([2, 1])

# 90-degree rotation matrix
rotation_matrix = np.array([
    [0, -1],
    [1,  0]
])

# Apply transformation
transformed_vector = np.dot(rotation_matrix, vector)

print("Original Vector:", vector)
print("Transformed Vector:", transformed_vector)
```

```python
# Create a 3-dimensional structure
arr = np.arange(1, 13)

# Reshape into 3 rows and 4 columns
matrix = arr.reshape(3, 4)

print("Original Array:\n", matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)

print("First Row:", matrix[0])
print("First Column:", matrix[:, 0])

print("Maximum:", np.max(matrix))
print("Minimum:", np.min(matrix))
print("Mean:", np.mean(matrix))
```
