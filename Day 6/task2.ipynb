```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Age": [22, 25, 28, 30, 35, 40, 45, 50, 55, 60],
    "Salary": [20000, 25000, 30000, 35000, 45000,
               50000, 60000, 65000, 75000, 90000],
    "Experience": [1, 2, 3, 5, 8, 10, 15, 18, 20, 25],
    "Performance": [60, 65, 70, 72, 78, 80, 85, 88, 90, 95]
}

df = pd.DataFrame(data)

df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Salary")
plt.title("Age vs Salary")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Salary"], kde=True)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Experience", y="Performance")
plt.title("Experience vs Performance")

plt.savefig("experience_performance.png", dpi=300, bbox_inches="tight")
plt.show()
```
