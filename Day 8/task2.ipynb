```python
from transformers import pipeline
import torch
```

```python
device = 0 if torch.cuda.is_available() else -1

print("Using GPU" if device == 0 else "Using CPU")
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2",
    device=device
)

print("Transformer model loaded successfully!")
```

```python
prompt = "Artificial Intelligence is transforming the world because"

print("Input Prompt:", prompt)
```

```python
result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7
)

print("Generated Text:")
print(result[0]["generated_text"])
```

```python
prompts = [
    "Machine learning is useful because",
    "The future of technology will",
    "Python programming helps developers"
]

for prompt in prompts:

    output = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7
    )

    print("\nPrompt:", prompt)
    print("Generated:", output[0]["generated_text"])
```

```python
prompt = input("Enter your text prompt: ")

output = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7
)

print("\nAI Generated Response:")
print(output[0]["generated_text"])
```
